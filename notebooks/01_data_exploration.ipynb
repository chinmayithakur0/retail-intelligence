```python
import pandas as pd
```

```python
# Load all supermarket price files
aldi = pd.read_csv('../data/raw/All_Data_Aldi.csv')
asda = pd.read_csv('../data/raw/All_Data_ASDA.csv')
morrisons = pd.read_csv('../data/raw/All_Data_Morrisons.csv')
sainsburys = pd.read_csv('../data/raw/All_Data_Sains.csv')
tesco = pd.read_csv('../data/raw/All_Data_Tesco.csv')

# Load demand data
demand = pd.read_csv('../data/raw/train.csv')
```

```python
# Check shapes
print("=== SUPERMARKET DATASETS ===")
print(f"Aldi:       {aldi.shape}")
print(f"ASDA:       {asda.shape}")
print(f"Morrisons:  {morrisons.shape}")
print(f"Sainsburys: {sainsburys.shape}")
print(f"Tesco:      {tesco.shape}")

print("\n=== DEMAND DATASET ===")
print(f"Demand: {demand.shape}")
```

```python
print("\n=== ALDI COLUMNS ===")
print(aldi.columns.tolist())
print(aldi.head(3))

print("\n=== DEMAND COLUMNS ===")
print(demand.columns.tolist())
print(demand.head(3))
```

```python
# Check date ranges
print("=== DATE RANGES ===")
print(f"Aldi dates:    {aldi['date'].min()} → {aldi['date'].max()}")
print(f"Tesco dates:   {tesco['date'].min()} → {tesco['date'].max()}")
print(f"Demand dates:  {demand['date'].min()} → {demand['date'].max()}")
```

```python
# Check categories in supermarket data
print("\n=== ALDI CATEGORIES ===")
print(aldi['category'].unique())
```

```python
# Check missing values
print("\n=== MISSING VALUES ===")
print("Aldi nulls:", aldi.isnull().sum().sum())
print("Demand nulls:", demand.isnull().sum().sum())
```

```python
# Check demand sales range
print("\n=== DEMAND SALES STATS ===")
print(demand['sales'].describe())
```
